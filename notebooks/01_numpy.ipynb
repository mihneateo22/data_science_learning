```python
#IMPORTANT (UP-CASTING)

# an array can only have one type. so when u give numpy mixed types, it picks the lowest kind that fits all inputs, then uses the default size (64-bit)
# bool  →  int  →  float  →  complex  →  string 
# a type from the right can represent everything on its left. FOR EXAMPLE:

import numpy as np

L = [3.14, 0, 1, 2, 3]
a = np.array(L) # --> np.array(L) will make every item in the array a float type.
print(a)
```

```python
#SOME TESTING

import numpy as np

print(np.array([True, 2, 3]), np.array([True, 2, 3]).dtype)
print()
print(np.array([1, 2, 3]) + 0.5, (np.array([1, 2, 3]) + 0.5).dtype)
print()
print(np.array([1, 2, 3]) / 1, (np.array([1, 2, 3]) / 1).dtype)
print()
print(np.array([1.9, 2.1], dtype=int))
```

```python
print((np.array([1, 2, 3]) // 2).dtype, np.array([1, 2, 3]) // 2)                           # dtype? values? dtype: int64, values : [0 1 1]
print(np.array([-1.9, 2.7], dtype=int))                                                     # values? (watch the negative) values : [-1 2]
print((np.array([1, 2, 3], dtype=np.int8) + np.array([1, 2, 3])).dtype)                     # dtype? dtype : int64
```

```python
#CREATING ARRAYS FROM SCRATCH. VERY IMPORTANT!!!!

print("length 10 integer array filled with 0 (zeros)")
print(np.zeros(10, dtype=int))
print()

print("3x5 floating point array filled with 1 (ones)")
print(np.ones(shape=(3, 5), dtype=float))
print()

print("4x4 array filled with pi (3.14)")
print(np.full((4,4), 3.14))
print()

print("Starting at 0, ending at 20, stepping by 2")
print(np.arange(0, 20, 2))
print()

print("array of five values evenly spaced between 0 and 1")
print(np.linspace(0, 1, 5))
print()

print("a 3x3 array of uniformly distributed, also random values between 0 and 1")
print(np.random.random((3, 3)))
print()

print("a 3x3 array of normally distributed random values with mean 0 and standard deviation 1")
print(np.random.normal(0, 1, (3, 3)))
print()

print("3x3 array of random integers in the interval [0, 10)")
print(np.random.randint(0, 10, (3, 3)))
print()

print("a 3x3 identity matrix")
print(np.eye(3))
print()
```
