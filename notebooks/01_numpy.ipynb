```python
# Definiți o funcție ce calculează norma Frobenius pentru o matrice A (i.e. implementati formula normei Frobenius). Comparați cu functia norm din Python numpy.linalg. 

import numpy as np

def main():
  elements = int(input("Enter the numbe of elements on row and column: "))
  print(f"Exercitiul 1 raspuns : ", norm_Frobenius(elements))


def norm_Frobenius(n):
  rng = np.random.default_rng()
  a = rng.random((n, n))
  print(a) #matricea generata random

  sum = 0
  for i in range(n):
    for j in range(n):
      sum += a[i, j] ** 2

  print(f" Norma Frobenius: {np.linalg.norm(a)}") #verificare daca am facut norma Frobenius corect
  return np.sqrt(sum)


if __name__ == "__main__":
  main()
```
